```python
import os
import dotenv
import pytidycensus as tc
import leafmap

dotenv.load_dotenv()
tc.set_census_api_key(os.environ["CENSUS_API_KEY"])
```

```python
vars = {
    "Total Population Estimate": "B01003_001"
}

geos = ["state", "county", "county subdivision", "tract", "block group"]

gdfs = {}
for geo in geos:
    gdf = tc.get_acs(
        geography=geo,
        variables=vars,
        state="RI",
        year=2024,
        geometry=True
    )
    gdfs[geo] = gdf
    gdf.to_file(f"../data/RIPopulationEstimate-{geo}-2024-acs5.geojson", driver="GeoJSON")
```
